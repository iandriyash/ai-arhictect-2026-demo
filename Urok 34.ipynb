```python
import random
from functools import reduce
import statistics
```

```python
def generate_random_list(a: int, b: int, length: int) -> list[int]:
    """
    Генерирует список случайных целых чисел из равномерного распределения.

    Parameters:
        a: нижняя граница диапазона (включительно).
        b: верхняя граница диапазона (включительно).
        length: длина результирующего списка.

    Returns:
        Список случайных целых чисел заданной длины.
    """
    if length <= 0:
        raise ValueError("Длина списка должна быть больше нуля")
    if a > b:
        raise ValueError("Нижняя граница 'a' не может быть больше верхней 'b'")
        
    return [random.randint(a, b) for _ in range(length)]
```

```python
def frequency_dict(lst: list[int]) -> dict:
    """
    Строит частотный ряд для значений списка с помощью функции высшего
    порядка reduce: на каждом шаге создаётся НОВЫЙ словарь, 
    а не изменяется старый — функция остаётся чистой.

    Parameters:
        lst: список чисел.

    Returns:
        Словарь {значение: частота}.
    """
    if not lst:
        return {}
    
    return reduce(
        lambda acc, x: {**acc, x: acc.get(x, 0) + 1},
        lst,
        {}
    )
```

```python
def calculate_stats(lst: list[int]) -> dict:
    """
    Считает базовую статистику по списку: минимум, среднее, медиану, моду, максимум.

    Parameters:
        lst: список чисел.

    Returns:
        Словарь с рассчитанными статистиками.
    """
    if not lst:
        raise ValueError("Невозможно рассчитать статистику для пустого списка")
    
    # Считаем моду через уже построенный частотный ряд (переиспользуем frequency_dict)
    freq = frequency_dict(lst)
    max_freq = max(freq.values())
    modes = [value for value, count in freq.items() if count == max_freq]
    mode_result = modes[0] if len(modes) == 1 else modes
    
    return {
        "min": min(lst),
        "mean": statistics.mean(lst),
        "median": statistics.median(lst),
        "mode": mode_result,
        "max": max(lst)
    }
```

```python
# --- Проверка работы функций ---
try:
    my_list = generate_random_list(1, 777, 888)
    print("Сгенерированный список:", my_list)
    print("Частотный ряд:", frequency_dict(my_list))
    print("Статистика:", calculate_stats(my_list))
except ValueError as e:
    print(f"Ошибка выполнения: {e}")
```
